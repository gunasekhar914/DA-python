dds one element to the set.

```python
#add
a= {10,20,30,40}
a.add(50)
print(a)
```

```python
#update
a= {10,20,30}
a.update([40,50,60])
print(a)
```

```python
a.update([70])
print(a)
```

```python
#remove
a= {10,20,30,0}
a.remove(20)
print(a)
```

```python
#discard
a= {10,20,30,40}
a.discard(20)
print(a)
```

```python
#pop
a.pop()
print(a)
```

```python
#clear
a={10,20,30,40}
a.clear()
print(a)
```

```python
#copy
a={10,20,30,40}
b= a.copy()
print(b)
```
